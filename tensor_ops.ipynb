```python
import torch
```

```python
#reshape and view
my_tensor = torch.arange(12)
my_tensor
```

```python
my_tensor = my_tensor.reshape(4,3)
my_tensor
```

```python
#Reshape if we don't know the number of items using -1
my_tensor2 = torch.arange(15)
my_tensor2
```

```python
my_tensor2 = my_tensor2.reshape(-1,3)
my_tensor2
```

```python
my_tensor3 = torch.arange(10)
my_tensor3
```

```python
my_tensor4 = my_tensor3.view(2,5)
my_tensor4
```
